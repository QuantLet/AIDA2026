## Output

`output/` holds 12 file(s) written by this stage.

````python
OUTPUT = HERE / 'output'
CSVS = ["chronos_t5_BTC.csv", "chronos_t5_BTC_N2000.csv", "chronos_t5_DAX.csv", "chronos_t5_DAX_N2000.csv", "chronos_t5_GOLD.csv", "chronos_t5_GOLD_N2000.csv", "chronos_t5_N225.csv", "chronos_t5_N225_N2000.csv", "chronos_t5_SPX.csv", "chronos_t5_SPX_N2000.csv", "chronos_t5_returns_SPX.csv", "chronos_t5_returns_SPX_facts.csv"]
summary = []
for f in CSVS:
    df = pd.read_csv(OUTPUT / f)
    summary.append({'file': f, 'rows': len(df),
                    'columns': ', '.join(df.columns[:6])})
pd.DataFrame(summary)
````
