## Output

`output/` holds 10 file(s) written by this stage.

````python
OUTPUT = HERE / 'output'
CSVS = ["headlines_BTC.csv", "headlines_DAX.csv", "headlines_GOLD.csv", "headlines_N225.csv", "headlines_SPX.csv", "coverage_BTC.csv", "coverage_DAX.csv", "coverage_GOLD.csv", "coverage_N225.csv", "coverage_SPX.csv"]
summary = []
for f in CSVS:
    df = pd.read_csv(OUTPUT / f)
    summary.append({'file': f, 'rows': len(df),
                    'columns': ', '.join(df.columns[:6])})
pd.DataFrame(summary)
````
