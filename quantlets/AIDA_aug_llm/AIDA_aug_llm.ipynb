## Output

````python
OUTPUT = HERE / 'output'
CSVS = ["llm_dated_BTC_claude-haiku-4-5.csv", "llm_dated_DAX_claude-haiku-4-5.csv", "llm_dated_GOLD_claude-haiku-4-5.csv", "llm_dated_N225_claude-haiku-4-5.csv", "llm_dated_SPX_claude-haiku-4-5.csv", "llm_dated_news_BTC_claude-haiku-4-5.csv", "llm_dated_news_DAX_claude-haiku-4-5.csv", "llm_dated_news_GOLD_claude-haiku-4-5.csv", "llm_dated_news_N225_claude-haiku-4-5.csv", "llm_dated_news_SPX_claude-haiku-4-5.csv", "llm_series_BTC_claude-haiku-4-5.csv", "llm_series_DAX_claude-haiku-4-5.csv", "llm_series_GOLD_claude-haiku-4-5.csv", "llm_series_N225_claude-haiku-4-5.csv", "llm_series_SPX_claude-haiku-4-5.csv", "llm_series_state_BTC_claude-haiku-4-5.csv", "llm_series_state_DAX_claude-haiku-4-5.csv", "llm_series_state_GOLD_claude-haiku-4-5.csv", "llm_series_state_N225_claude-haiku-4-5.csv", "llm_series_state_SPX_claude-haiku-4-5.csv"]
summary = []
for f in CSVS:
    df = pd.read_csv(OUTPUT / f)
    summary.append({'file': f, 'rows': len(df),
                    'columns': ', '.join(df.columns[:6])})
pd.DataFrame(summary)
````
