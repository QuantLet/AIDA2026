## Output

`output/` holds 10 file(s) written by this stage.

````python
OUTPUT = HERE / 'output'
CSVS = ["returns_BTC.csv", "returns_DAX.csv", "returns_GOLD.csv", "returns_N225.csv", "returns_SPX.csv", "bench_BTC.csv", "bench_DAX.csv", "bench_GOLD.csv", "bench_N225.csv", "bench_SPX.csv"]
summary = []
for f in CSVS:
    df = pd.read_csv(OUTPUT / f)
    summary.append({'file': f, 'rows': len(df),
                    'columns': ', '.join(df.columns[:6])})
pd.DataFrame(summary)
````
